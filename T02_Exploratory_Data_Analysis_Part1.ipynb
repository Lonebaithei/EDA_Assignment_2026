Step 6: Visualise Numerical Variables

```python
plt.hist(orders['days_since_prior_order'].dropna(), bins=30)
plt.xlabel("Days Since Prior Order")
plt.ylabel("Frequency")
plt.title("Distribution of Days Between Orders")
plt.show()
```

Step 7: Repeat for Additional Numerical Variables

```python
# Variable 2 (orders_hour_of_day)
orders['order_hour_of_day'].describe()
```

```python
# Variable 3 (order_number)
plt.hist(orders['order_hour_of_day'], bins=24)
plt.xlabel("Hour of Day")
plt.ylabel("Frequency")
plt.title("Orders by Hour of Day")
plt.show()
```

Step 8: Univariate Analysis of Categorical Variables

```python
# Categorical Variable 1: order_dow
orders['order_dow'].value_counts()
```

Step 9: Visualise Categorical Variables

```python
orders['order_dow'].value_counts().sort_index().plot(kind='bar')
plt.xlabel("Day of Week")
plt.ylabel("Number of Orders")
plt.title("Orders by Day of Week")
plt.show()
```
