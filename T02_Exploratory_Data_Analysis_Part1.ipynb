Import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Load INTERIM Data

```python
orders = pd.read_csv("../data/interim/orders_interim.csv")
products = pd.read_csv("../data/interim/products_interim.csv")
order_products = pd.read_csv("../data/interim/order_products_interim.csv")
```

Univariate Numerical Analysis

```python
orders['days_since_prior_order'].describe()
```

Interpretation

The median number of days between orders appears to be approximately X days, suggesting that customers typically reorder within this time frame.

The wide range indicates variability in purchasing behaviour.

The presence of higher values suggests some customers reorder infrequently.

Visualisation of it

```python
plt.hist(orders['days_since_prior_order'].dropna(), bins=30)
plt.xlabel("Days Since Prior Order")
plt.ylabel("Frequency")
plt.title("Distribution of Days Between Orders")
plt.show()
```

Additional Numerical Variables

```python
# Variable 2 (orders_hour_of_day)
orders['order_hour_of_day'].describe()
```

```python
# Variable 3 (order_number)
plt.hist(orders['order_hour_of_day'], bins=24)
plt.xlabel("Hour of Day")
plt.ylabel("Frequency")
plt.title("Orders by Hour of Day")
plt.show()
```

Categorical Variable Analysis

```python
# Categorical Variable 1: order_dow
orders['order_dow'].value_counts()
```

Visualization

```python
orders['order_dow'].value_counts().sort_index().plot(kind='bar')
plt.xlabel("Day of Week")
plt.ylabel("Number of Orders")
plt.title("Orders by Day of Week")
plt.show()
```

```python
# Categorical Variable 2: Product Frequency
top_products = order_products['product_id'].value_counts().head(10)
top_products
```
