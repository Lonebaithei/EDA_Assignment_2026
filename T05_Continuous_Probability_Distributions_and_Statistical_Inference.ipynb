**B2. Feature Engineering: Order Size**

```python
ordersize = order_products.groupby('order_id').size ()
```


*   Compute mean and standard deviation.
*   Plot histogram.
*   Assess normality.


Computation of the **Mean** and **Standard Deviation**.

```python
mean_ordersize = ordersize.mean()
std_ordersize = ordersize.std()

print(f"Mean order size: {mean_ordersize:.2f}")
print(f"Standard deviation of order size: {std_ordersize:.2f}")
```

Plotting a **Histogram** showing the distribution of **order_size**.

```python
plt.figure(figsize=(10, 6))
plt.hist(ordersize, bins=50, edgecolor='black', alpha=0.7)
plt.title('Distribution of Order Size')
plt.xlabel('Number of Products in an Order')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

**Interpretation**

The mean order size is approximately 10.09, and the standard deviation is approximately 7.53. **This indicates that on average, an order contains about 10 products, with a moderate spread around this average**.

### Assessing Normality: Skewness and Kurtosis

```python
skewness = skew(ordersize)
kurt = kurtosis(ordersize)

print(f"Skewness of order size: {skewness:.2f}")
print(f"Kurtosis of order size: {kurt:.2f}")

if abs(skewness) < 0.5 and abs(kurt) < 0.5:
    print("The distribution appears approximately normal based on skewness and kurtosis.")
elif skewness > 0.5:
    print("The distribution is positively skewed (has a long tail to the right).")
elif skewness < -0.5:
    print("The distribution is negatively skewed (has a long tail to the left).")

if kurt > 0.5:
    print("The distribution is leptokurtic (has heavier tails and a sharper peak than a normal distribution).")
elif kurt < -0.5:
    print("The distribution is platykurtic (has lighter tails and a flatter peak than a normal distribution).")
```

### Assessing Normality: Q-Q Plot

```python
sm.qqplot(ordersize, line='s')
plt.title('Q-Q Plot of Order Size')
plt.show()
```

**B3. Z-Score Analysis**

*   Identify observations with |Z| ≥ 3.
*   Interpret unusual values.

```python
z_scores = (ordersize - mean_ordersize) / std_ordersize

outliers = ordersize[abs(z_scores) >= 3]

print(f"Number of observations with |Z| >= 3: {len(outliers)}")
print("Observations with |Z| >= 3 (first 10 if many):")
print(outliers.head(10))
```

**Interpretation of unusual values**

*   The Z-score analysis has been completed and we found 48,066 observations where the absolute Z-score was greater than or equal to 3.

*   These observations represent orders with an unusually high or low number of products compared to the average.
*   The output shows the order_id and the ordersize (number of products) for the first 10 of these potential outliers. For instance, order_id 106 had 34 products, order_id 742 had 45 products, which are significantly larger than the mean order size of approximately 10 products.

*    This indicates that these orders are quite large and deviate substantially from the typical order size.
